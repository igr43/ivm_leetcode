# 49. Group Anagrams

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:

        result = {}
        for word in strs:
            key = [0] * 26
            for char in word:
                
                key[ord(char) - ord('a')] += 1
            key = tuple(key)
            if key not in result:
                result[key] = []
            result[key].append(word)
        return list(result.values())
```

# 347. Top K Frequent Elements

```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:

        count = {}
        frequency = [[] for _ in range(len(nums) + 1)]

        for num in nums:
            count[num] = 1 + count.get(num, 0)

        for num, c in count.items():
            frequency[c].append(num)

        result = []
        for i in range(len(frequency) - 1, 0, -1):
            for num in frequency[i]:
                result.append(num)
                if len(result) == k:
                    return result
                
```

# 128. Longest Consecutive Sequence

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:

        nums_set = set(nums)
        longest = 0
        for num in nums_set:
            if (num-1) not in nums_set:

                seq_len = 0
                while (num + seq_len) in nums_set:
                    seq_len += 1
                longest = max(seq_len, longest)

        return longest
```

# 704. Binary Search


```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:

        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] > target:
                right = mid - 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                return mid
            
        return -1
```

# 33. Search in Rotated Sorted Array

```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:

        left = 0
        right = len(nums) - 1 
        while left <= right:
            mid = (left+right) // 2

            middle_value = nums[mid]
            if middle_value == target:
                return mid
            
            if nums[left] <= middle_value:
                # левая часть отсортирована 
                if target > middle_value or target < nums[left]:
                    left = mid + 1
                else:
                    right = mid - 1

            else:
                # правая часть отсортирована  
                if target < middle_value or target > nums[right]:
                    right = mid - 1
                else:
                    left = mid + 1

        return -1
```
