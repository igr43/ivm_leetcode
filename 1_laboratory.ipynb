# 1. Two Sum

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        viewed_values = dict()
        for idx, num in enumerate(nums):
            difference = target - num 
            if difference in viewed_values:
                return [viewed_values[difference], idx]
        
            viewed_values[num] = idx

        
```
