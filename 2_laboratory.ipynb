# 20. Valid Parentheses

```python
class Solution(object):
    def isValid(self, s):
        """
        :type s: str
        :rtype: bool
        """
        close_to_open = {
            ")" : "(",
            "]" : "[",
            "}" : "{",
        }

        stack = []

        for c in s:
            if c in close_to_open:
                if not((stack) and (close_to_open[c] == stack.pop())):
                    return False
            else:
                stack.append(c)

        return not stack
```
