# 20. Valid Parentheses

```python
class Solution(object):
    def isValid(self, s):
        """
        :type s: str
        :rtype: bool
        """
        close_to_open = {
            ")" : "(",
            "]" : "[",
            "}" : "{",
        }

        stack = []

        for c in s:
            if c in close_to_open:
                if not((stack) and (close_to_open[c] == stack.pop())):
                    return False
            else:
                stack.append(c)

        return not stack
```

# 150. Evaluate Reverse Polish Notation 

```python
class Solution(object):
    def evalRPN(self, tokens):
        """
        :type tokens: List[str]
        :rtype: int
        """
        stack = []
        for token in tokens:
            if token == '+':
                stack.append(stack.pop() + stack.pop())
            elif token == '*':
                stack.append(stack.pop() * stack.pop())
            elif token == '-':
                a = stack.pop()
                b = stack.pop()
                stack.append(b - a)
            elif token == '/':
                a = stack.pop()
                b = stack.pop()
                stack.append(int(b / a))
            else:
                stack.append(int(token))
        return stack[0]
            
```
